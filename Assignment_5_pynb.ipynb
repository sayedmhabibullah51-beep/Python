```python
def process_scores(scores, threshold):
  above = []
  below_or_equal = []

  for score in scores:
    if score > threshold:
      above.append(score)
    else:
      below_or_equal.append(score)

  return above,below_or_equal

scores= [45, 67, 89, 32, 76, 50,43,90,454,22,11,65]
threshold= 50

above_score,below_or_equal= process_scores(scores, threshold)

print('Above Score :',above_score)
print('Below Or Equal Score :',below_or_equal)
```

```python
def convert_temp(value, to_unit="C"):
  if to_unit == 'C':
    return (value - 32)* 5/9
  elif to_unit == 'F':
    return (value * 9 / 5) + 32
  else:
    return 'Invalid unit'

print('Using  Positional Argument')
print('Fahrenheit to Celsius :',convert_temp(98.6, 'C'))
print('Celsius to Fahrenheit :',convert_temp(37, 'F'))

print('\nUsing keyword Argument')
print('Fahrenheit to Celsius :',convert_temp(value=212, to_unit='C'))
print('Celsius to Fahrenheit :',convert_temp(value=0, to_unit='F'))
```

```python
def safe_divide(a,b):
  try:
    result= a/b
    return {'Success': True,'Result': result}
  except ZeroDivisionError:
    return{'Success':False,'Error':'Division by zero is not allowed.'}
  except TypeError:
    return{'Success':False,'Error':'Invalid input type. Both a and b must be numbers.'}

print(safe_divide(4,2))
print(safe_divide(4,0))
print(safe_divide('a',2))
print(safe_divide(4,'b'))
```

```python
class BankAccount():
  def __init__(self,owner,balance):
    self.owner= owner
    self.balance= balance
  def deposite(self,amount):
    if amount > 0:
      self.balance += amount
      print(f'Deposited : ${amount:.2f}')
    else:
      print('Deposited Amount Must Be Positive')
  def withdraw(self,amount):
    if self.balance < amount:
      print('Incuficient Funds! Withdrawal Canceled.')
    elif amount <=0:
      print('Withdrawal Amount Must Be Positive')
    else:
      self.balance -= amount
      print(f'Withdrawn: ${amount:.2f}')
  def get_balance(self):
    print(f'\nAccount Name: {self.owner}\nBalance: ${self.balance:.2f}\n')

account = BankAccount('Shanto', 1000)

account.deposite(50)
account.get_balance()
account.withdraw(30)
account.withdraw(1150)
account.deposite(300)
account.get_balance()



```

```python
class Student():
  def __init__(self,name,marks):
     self.name= name
     self.marks= marks

  def average(self):
    total= sum(self.marks.values())
    return total / len(self.marks)

student1 = Student(
    "Shanto",
    {"Math": 85, "Physics": 98, "Chemistry": 90}
)

student2 = Student(
    "Pranto",
    {"Math": 72, "English": 88, "Biology": 80}
)

print(f"{student1.name}'s average score : {student1.average():.2f}")
print(f"{student2.name}'s average score : {student2.average():.2f}")
```

```python
import os
from datetime import datetime


class InventoryReport:
    def __init__(self, item_name, stock):
        self.item_name = item_name
        self.stock = stock


def generate_inventory_summary(report_obj, filename="inventory_report.txt", header="INVENTORY SUMMARY"):
    try:
        if not isinstance(report_obj, InventoryReport):
            raise TypeError("Invalid object type. Expected InventoryReport.")

        total_stock = sum(report_obj.stock.values())

        current_time = datetime.now().strftime("%d-%m-%Y %I:%M:%S %p")

        with open(filename, "w") as file:
            file.write(f"{header}\n")
            file.write(f"Generated on: {current_time}\n\n")
            file.write(f"Item Name: {report_obj.item_name}\n")
            file.write("Stock Details:\n")

            for location, count in report_obj.stock.items():
                file.write(f"  {location}: {count}\n")

            file.write(f"\nTotal Stock: {total_stock}\n")

        return os.path.abspath(filename)

    except Exception as e:
        print(f"Error: {e}")


inventory = InventoryReport(
    "Laptop",
    {
        "Warehouse A": 50,
        "Warehouse B": 30,
        "Store Front": 20
    }
)

path1 = generate_inventory_summary(inventory, "report1.txt", "WAREHOUSE REPORT")
print("Report saved at:", path1)

path2 = generate_inventory_summary(
    report_obj=inventory,
    filename="report2.txt",
    header="INVENTORY STATUS"
)
print("Report saved at:", path2)

generate_inventory_summary("NotAnInventoryObject")
```
