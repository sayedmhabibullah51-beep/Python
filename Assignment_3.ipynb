SECTION C — Dictionaries & Methods

Q5

```python
sales = {
    "Jan": {"total": 1200, "returns": 2},
    "Feb": {"total": 1500, "returns": 3},
    "Mar": {"total": 1700, "returns": 1}
}

for month,data in sales.items():
  data['total']=round(data['total']*1.05,2)
print('After increasing by 5% every month :',sales)

sales.update({"Apr": {"total": 1600, "returns": 2}})
print('After adding April :',sales)

max_returns_month=max(sales,key=lambda m:sales[m]['returns'])
del sales[max_returns_month]
print('After deleting month with higest returns :',sales)

above_1500=[]
for month, info in sales.items():
  if info['total'] > 1500:
     above_1500.append(month)
print('Month with total > 1500 :',above_1500)
```

Q6

```python
students = {
    1: {"name": "Riya", "marks": {"math": 88, "eng": 76, "sci": 90}},
    2: {"name": "Fahim", "marks": {"math": 92, "eng": 81, "sci": 79}},
    3: {"name": "Simran", "marks": {"math": 67, "eng": 72, "sci": 69}}
}

for id, mark in students.items():
  mark['marks'].update({"computer": 85})
  print('After adding mark of computer:85 in every students marks :',mark)

for id,data in students.items():
  data['total']=sum(data['marks'].values())
print('After calculating total scores and adding :',students)

max_total=max(data['total'] for data in students.values())

for id,data in students.items():
  if data['total'] == max_total:
    print('Student with highest total :',data['name'])

for id,info in students.items():
  marks=info['marks']
  if all(mark>70 for mark in marks.values()):
    print('Students who have got above 70 in all subject :',info['name'])
```

```python
products = {"p1": 120, "p2": 250, "p3": 80, "p4": 300}
print('The products whose price is between 100 and 260 :')
for product,price in products.items():
  if 100<price<260:
    print(product ,end=' ')

new_products={}
for product,price in products.items():
  if price>200:
    new_products[product]='High'
  else:
    new_products[product]='Low'
print('\nCreating a new Dictionary :',new_products)
```

```python
names = ["Amit", "Sara", "Rohan", "Kia"]
scores = [88, 92, 75, 95]
grades = ["B", "A", "C", "A"]

students=list(zip(names,scores,grades))

print('Students with grade "A" :',end = ' ')
for names,scores,grades in students:
  if grades == 'A':
    print(names,end = ' ')

```

```python
inventory = {
    "fruits": {"apple": 50, "banana": 30},
    "vegetables": {"carrot": 40, "spinach": 25},
    "drinks": {"water": 100, "juice": 40}
}

for category in inventory:
  for item in inventory[category]:
    inventory[category][item]= int(inventory[category][item]*1.10)
print('After increase quantity of every item by 10% :',inventory)


new_inventory = {}

for category,items in inventory.items():
  temp={}
  for item,quantity in items.items():
    if quantity>40:
      temp[item]=quantity
  if temp:
    new_inventory[category]=temp

print('Items whose updated quantity > 40 :',new_inventory)


```
