Simple Calculator

```python
# It will take the first and second input from user

First=input('Enter First Number:')
Second=input('Enter Second Number:')

#Convert both inputs into integers
First=int(First)
Second=int(Second)

# Here i calculated Addition, Subtraction,Multiplication,Division
Addition=First+Second
Subtraction=First-Second
Multiplication=First*Second
Division=First/Second

# Here i printed Addition, Subtraction,Multiplication,Division
print('Addition:',Addition)
print('Subtraction:',Subtraction)
print('Multiplication:',Multiplication)
print('Division:',Division)



```

Temperature Converter

```python
#Turning Celsius into type
C = input('Enter the temperature in Celsius:')
#Converted to float
C = float (C)
#Writing the formula
F = (C * 9/5) + 32
#Using f string for Fahrenheit
print(f'Temperature in Fahrenheit is: {F}°F')
```

Billing Program

```python
#Turning item name, price, quantity in input type
Item_name=input('Enter the item name:')
Price=input('Enter the item price:')
Quantity=input('Enter the item quantity:')

#Turning price and quantity in int
Price=int(Price)
Quantity=int(Quantity)

#Using the formula
Total = Price * Quantity

#Using f string for printing
print(f'Total Bill for {Item_name} = {Total}')


```
