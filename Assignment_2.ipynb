```python
# 5 + 3 * 4
# Multiplication happens before addition
print(5 + 3 * 4) # 3*4 = 12 then 5 + 12 = 17

# (5 + 3) * 4
# Addition happens first because of Parentheses
print((5 + 3) * 4) # (5+3=8) then 8*4=32

# 100 / 6 and 100 // 6
# Division and Floor division
print(100/6) #  16.666666666666668 printing all the floor number
print(100//6) # 16 does not printing the floor number

# 2 ** 5 * 3
# Exponention happens first then addition happen
print(2 ** 5 * 3) # 2**5=32 then 32*3=96
```

Q10. List Methods

```python
from os import remove
nums = [12, 4, 8, 1, 9, 3, 4, 4]
nums.sort
print('sort:',nums)
nums[2]=100
print('Inserting 100 at index 2:',nums)
nums.remove(1)
print('Remove the value 1:',nums)
print('Max value:',max(nums))
print('Min value:',min(nums))
nums.reverse()
print('Reversing the list:',nums)
print('Counting the times of 4 appears:',nums.count(4))
```

Q11. List Processing

```python
integer_list=[55,44,66,88,99,5,10,2,29,100]
print('Printing the list:',integer_list)

odd_list=[i for i in integer_list if i%2!=0]
print('Printing the odd numbers:',odd_list)

greater_list=[i for i in integer_list if i>20]
print('Printing the numbers greater than 20:',greater_list)

lesser_list=[i if i>10 else 0 for i in integer_list]
print('Printing the numbers lesser than 10:',lesser_list)
```
